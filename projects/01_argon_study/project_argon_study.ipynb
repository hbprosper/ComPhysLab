## Introduction
This project is inspired by the beautiful results by Rahman [1], who simulated a box of argon atoms in 1964. In this project you will simulate a microscopic spherical container of argon atoms and study the behavior of this system. We assume that force on atom $i$ due to atom $j$ can be modeled using the **Lennard-Jones** (LJ) force [2,3] given by
\begin{align}
    \mathbf{F}_i(r) & = 24 \left(\frac{\epsilon}{\sigma}\right)\left[ 2 \left(\frac{\sigma}{r}\right)^{13} - \left(\frac{\sigma}{r}\right)^7\right] \hat{r}, \text{ where}\nonumber\\
    \hat{r} & = \frac{\mathbf{r}_i - \mathbf{r}_j}{r},\tag{3}
\end{align}
is a unit vector that points from atom $j$ to atom $i$. The quantity $r = ||\mathbf{r}_i - \mathbf{r}_j||$  is the distance between the atoms, where $\mathbf{r}_i$ and $\mathbf{r}_j$ are the position vectors of atoms $i$ and $j$, respectively; 
$\epsilon$ is an energy that characterizes the interaction and $\sigma$ is the distance at which (by convention) $V(r) = 0$. By the same convention, $V(r)$ is also zero at $r = \infty$.
The first term in the potential yields a **repulsive force**, while the second term yields an **attractive force**. Because the potential falls off rapidly with the distance $r$ between atoms the latter are essentially free particles when their separation is much greater than $\sigma$. 

**Argon Parameters**

The parameters for argon in SI units are:
  * $\sigma = 3.4 \times 10^{-10} \text{ m}$;
  * $\epsilon =  1.657 \times 10^{-21} \text{ J}$;
  * $m = 6.69\times 10^{-26} \text{ kg}$, and
  * $\rho_0 = 1.642 \text{ kg/m}^3$ @ 293 K and one atmosphere.

In practice, we choose units so that the Newton's second law for this problem can be written as
\begin{align}
    24 \left[ 2 \left(\frac{1}{r}\right)^{13} - \left(\frac{1}{r}\right)^7\right] \hat{r} & = \frac{d\mathbf{v}}{dt}.
\end{align}

## A little bit of physics

### Kinetic energy and temperature

In Newtonian physics the kinetic energy of a particle of mass $m$ and velocity $\vec{v}$ is given $\frac{1}{2} m v^2$. The classical **equipartition theorem** implies that $$\frac{3}{2} k T = \frac{1}{2} m \overline{v^2},$$ where  $\overline{v^2}$ denotes the average of the square of the velocities for all particles in the system. Therefore, for this project, we shall *define* the temperature using 
\begin{align}
    T \equiv \frac{m}{k}\frac{\overline{v^2}}{3}.
\end{align}
However, since you'll work in units in which $m = k = 1$, the (unitless) temperature is given by
\begin{align}
    T \equiv \frac{\overline{v^2}}{3}.
\end{align}
To convert the unitless temperature to the temperature in kelvin, you need to multiply the above by the `T2K` conversion factor in the bag.

There are many ways to characterize a collection of particles. We consider two ways, below.

### Radial density function

For a gas we expect the distribution of the position of the atoms to exhibit no discernable structure because we expect the particles to be uniformly distributed on average. Consider a collection of $N$ atoms and let's focus on a single atom. Consider a shell, centered on the atom, of inner radius $r$ and outer radius $r+\Delta r$. If the *number* density is $n$, that is, the number of particles per unit volume, then the number of atoms in the shell is given by
\begin{align}
    \delta N & = n \, \frac{4}{3}\pi \left((r + \Delta r)^3 - r^3 \right).
\end{align}
Therefore, this is the total number of radial distances from the given atom to atoms within the shell. But since there are $N$ atoms, the total number of radial distances within shells of inner radius $r$ and thickness $\Delta r$ about every atom is
\begin{align}
    \Delta N & = N \, \delta N.
\end{align}
If there is structure in the distribution of the atoms, the average counts, $\Delta C$, within the shells will differ from $\Delta N$. The (binned) **radial density function** is defined as
\begin{align}
    g(r) & = \frac{\Delta C}{\Delta N},
\end{align}
which in the idealized limit that the shells are infinitely thin becomes
the **derivative** $g(r) = dC / dN$.

If the atoms in a collection of atoms are uniformly distributed, then we expect $g(r) = 1$ for all values of $r$. In this project, however, the atoms are confined to a microscopic spherical container, so we may be deviations from $g(r) = 1$. But, presumably, for a gas there should be no structure in the distribution.

### Maxwell distribution 

The following is one of the most profoundly useful formulas in the whole of modern physics,
\begin{align}
    dG & = g \frac{\Delta \Gamma}{h^3}.
\end{align}
It computes the number of quantum states, $dG$, in a system, within the momentum range $p$ to $p + dp$,
where $\Delta\Gamma = V (4 \pi p^2 dp)$ is called the **classical phase space volume**, $V$ is the volume of the enclosure containing the particle, $p$ is the magnitude of the particle's momentum, and $g$ is the **degeneracy factor**. For a given energy level (or, equivalently, momentum level) the number $g$ specifies the number of quantum states at that level. The number $h$ is Planck's constant. This beautiful formula connects classical and quantum physics and is a foundational expression in **statistical mechanics**.

If, in addition, we're given the probability (or probability density), $f(E)$, for a particle to occupy a set of quantum states around the energy $E$ and we know the number of quantum states available, $dG$, then the number of particles, $dN$, occupying these states is given by the chance of occupying a state times the number of states available, that is,
\begin{align}
    dN  & = f(E) \, dG,\\
        & = f(E) \, g \frac{V (4 \pi p^2 dp)}{h^3} .
\end{align}

For many systems, $f(E) = A \exp(-E / (kT)$, where $k$ is the Boltzmann constant and $T$ is the **absolute temperature** of the system of particles. For slowly moving particles $E = \frac{1}{2} m v^2$. The constant $A$ can be determined by imposing the condition $\int dN = N$. That is, if we integrate over all possible *occupied* energy states we should get the total number of  particles in the system. Therefore, we can write
\begin{align}
    dN  & = A e^{- m v^2 / (2 kT)} p^2 dp,\\
        & = A e^{- m v^2 / (2 kT)} v^2 dv,
\end{align}
where all the constants have been absorbed into $A$ and where for neutral argon, which we treat as a structureless particle, $g = 1$. Defining, $x = \sqrt{m/(kT)} v$, we can write
\begin{align}
    dN  & = A e^{- m v^2 / (2 kT)} p^2 dp,\\
        & = A e^{- x^2/ 2} x^2 dx ,
\end{align}
where, again, all constants have been absorbed into $A$.
Performing the integral $\int_0^\infty dN = N$, yields $A(\pi/2)^{1/2} = N$, hence $A = N (2/\pi)^{1/2}$. Thus do we arrive Maxwell's celebrated velocity (strictly speaking speed) distribution, 
\begin{align}
    dN  & = N (2/\pi)^{1/2} e^{- x^2/ 2} x^2 dx ,\\
        & = N \left(\frac{2}{\pi}\right)^{1/2}\left(\frac{m}{kT}\right)^{3/2} e^{- m v^2/(2 kT)} v^2 dv,\\
        & = N 4 \pi  \left(\frac{m}{2 \pi kT}\right)^{3/2} e^{- m v^2/(2 kT)} v^2 dv .
\end{align}
The derivative $f(v) = dN/dv$ is the **Maxwell velocity density function**. The quantity $f(v) dv$ gives  the average number of particles with speed between $v$ and $v + dv$.

## PART 1: Simulation

```python
def define_initial_state(rho, T, filename):
    # -----------------------------------------------------
    # Create a logger to log simulation data to file
    # -----------------------------------------------------
    # These quantities are written out once
    attributes = [
        'mass',      # Mass of argon atom (kg)
        'epsilon',   # Characteristic energy (J)
        'tc',        # Characteristic time scale (s)
        'vc',        # Characteristic speed (m/s)
        'sigma',     # Characteristic distance scale (m)
        'equi_sep',  # Equilibrium separation (sigma)
    
        'rho',       # Number density (number / sigma^3)
        'T',         # Temperature (K)
    
        'N',         # Number of atoms
        'R',         # Radius of container
        'rmin_sep',  # Minimum separation of atoms
        
        'dt',        # Simulation time step (tc)
        
        'T2K',       # Conversion to kelvin
        'vrms'
    ]
    
    # These quantities are written out periodically
    datasets = [
        'r',         # Positions of atoms   (in units of sigma)
        'v',         # Velocities of atoms  (in units of vc)
        'impulse',   # Momentum change imparted to container wall (mass*vc)
        'U'          # Potential energy of atoms (in units of epsilon)
    ]
    
    bg = argon_initial_state(rho, T)

    bg.save_every = 10  # Save every 10 frames
    bg.frame_rate = 30  # Frames/second
    bg.duration   = 80  # Duration of animation (in clock time) (s)
    bg.max_frame  = bg.duration * bg.save_every * bg.frame_rate

    print(f'''
        Number of time steps:        {bg.max_frame:10d}
        Save every:                  {bg.save_every:10d} time steps
        Number of saved frames:      {bg.frame_rate * bg.duration:10d}
    ''')

    bg.logger = SimLogger(
        filename, bg, 
        attributes, datasets, 
        bg.save_every)
    
    return bg
```

## Propagation and bounce functions

```python
def propagate(bg):
    # -----------------------------------------------------------
    # Update state of every particle.
    # -----------------------------------------------------------
    bg.r0[:] = bg.r # copy r into r0 before update
    bg.v0[:] = bg.v # copy v into v0 before update
    bg.r[:], bg.v[:], bg.U = propagate_order3(
        bg.k, bg.q, bg.m, bg.r0, bg.v0, bg.law, bg.dt)
    
def bounce_off_wall(bg, eps=1e-12):
    """
    Elastic reflection from inner wall of sphere 
    of radius bg.R centered at the origin.
    """

    # Detect which predicted positions are outside sphere
    r_mag  = magnitude(bg.r)                 # (n,)
    bounce = r_mag >= bg.R                   # (n,)
    if not np.any(bounce):
        bg.impulse = 0.0
        return

    # Pick subset of atoms to bounce
    r0 = bg.r0[bounce]                       # (k,3)
    v0 = bg.v0[bounce]                       # (k,3)
    r  = bg.r[bounce]                        # (k,3)
    v  = bg.v[bounce]                        # (k,3)

    # Direction of travel at start of step 
    # (for ray/sphere intersection)
    u0 = unit(v0)                            # (k,3)

    # Bounce points on the sphere
    # b: (k,3) points on |b|=R
    b, _, _ = line_sphere_intersect(r0, u0, bg.R)  

    # Outward normal at hit points
    n = unit(b)                              # (k,3)

    # Normal speed at impact (outward positive). 
    # Should be > 0.
    vdotn = dot(v0, n)                       # (k,)

    # Reflect velocity direction about normal: v' = v - 2(v·n)n
    v_reflected = v0 - 2 * vdotn[:, None] * n    # (k,3)

    # Compute impulse (change in momentum) delivered to wall 
    # (assuming unit mass). This is just the sum of the 
    # magnitudes of the changes in momentum.
    # Protect against possible negative values due to floating 
    # point rounding errors.
    bg.impulse = 2.0 * float(np.sum(np.clip(vdotn, 0, None)))
    
    # Use predicted speed
    speed = magnitude(v)                     # (k,)

    # New velocities after bounce
    bg.v[bounce] = speed[:, None] * unit(v_reflected)

    # Move positions: keep "remaining distance" from bounce point, 
    # but along reflected direction.
    # Remaining distances traveled during step:
    s = magnitude(r - b)                     # (k,)

    # Nudge slightly inward along normal just in case of rounding
    # errors.
    b_in = b - eps * n

    bg.r[bounce] = b_in + s[:, None] * unit(v_reflected)
```

## Define scene

```python
def build_scene(bg):
    # ----------------------------------------------------
    # Required attributes
    # ----------------------------------------------------
    bg.rate   = 30         # No faster than "rate" frames/second
    bg.frame  = 0          # Frame counter
    bg.active = True       # Controled by Stop button 
    bg.update = False      # Controled by Start/Pause button
    
    # ----------------------------------------------------
    # Create empty scene
    # ----------------------------------------------------
    # All scene widgets must be placed in bag.gfx so that they
    # can be properly deleted when the animation is stopped.
    gfx = bg.gfx           # give bag.gfx a shorter name

    bg.height = 300
    bg.size   = bg.R       # Scale of scene (units of sigma)
    gfx.scene = Scene(
        'Argon Atoms\n', bg.size, up=K, height=bg.height)
    gfx.scene.forward = vp.vector(-1,0,0)
    
    # ----------------------------------------------------
    # Add scene elements
    # ----------------------------------------------------
    gfx.container = vp.sphere(
        pos=vp.vector(0.0,0.0,0.0), 
        radius=bg.R, color=vp.color.cyan, opacity=0.05)

    # Model each atom as a small sphere
    gfx.atoms = []
    atom_radius = 0.02 * bg.R
    
    for i in range(bg.N):
        gfx.atoms.append(vp.sphere(
            pos=vp.vector(*bg.r[i]), 
            color=vp.color.yellow,
            radius=atom_radius))

    # Add control buttons (Stop, Start/Pause)
    controls = Controls(bg)

    gfx.b_stop  = vp.button(
        text="Stop",
        background=vp.color.red,
        pos=gfx.scene.title_anchor,
        bind=controls.stop)

    gfx.b_start_pause = vp.button(
        text="Start",
        background=vp.color.green,
        pos=gfx.scene.title_anchor,
        bind=controls.start_pause)

    # Add zoom slider
    gfx.zoom = Zoom(gfx.scene)
    
    # ------------------------------------------------------
    # Add a label to display the date 
    # ------------------------------------------------------
    # We want the label to remain in a fixed position, so we
    # need to work in screen space, not world space.
    # Note: In vpython, the bottom-left corner is the
    # (x,y) origin of the pixel coordinates (0,0,0).

    gfx.frame_label = vp.label(
        text=f'frame: {1:10d}',
        screen=True,              # Make label in screen-space, not world-space
        pixel_pos=True,           # Use pixel coordinates, not world coordinates
        pos=vp.vector(10, 20, 0), # Pixels from bottom-left: (x, y)
        align='left',             # pos pertains to label left
        height=16,                # Font size in pixels
        color=vp.color.white,
        opacity=0,                # 0 = transparent background box
        border=0,
        box=False,
        line=False
    )
```

```python
def update(bg):

    # Solve Newton's 2nd for all atoms
    propagate(bg)

    # Handle bounces off container wall
    bounce_off_wall(bg)

    # Log simulation data to file  every 'bg.save_every' frames
    # Note: This handled internally by the write function.
    bg.logger.write(bg)
    
    if bg.frame >= bg.max_frame:
        bg.active = False # stop animation
    
    # Loop over graphical objects and update their positions.
    for i, atom in enumerate(bg.gfx.atoms):
        x, y, z = bg.r[i]
        atom.pos.x = x
        atom.pos.y = y
        atom.pos.z = z

    # Update frame counter display
    if bg.frame % bg.rate == 0:
        bg.gfx.frame_label.text  = f'frame: {bg.frame:10d}',
```

## Run simulation without or with animation

I recommend running the simulation without the animation. Use the animation only to check that things look reasonable.

```python
def simulate(bg):

    start_time = time()
    
    for step in range(bg.max_frame):
        
        if step % 100 == 0:
            print(f'\r{step:10d}\t{ctime()}', end='')
    
        # Solve Newton's 2nd law for all particles 
        # in current time step, dt
        propagate(bg)
    
        # Handle bounces off container wall
        bounce_off_wall(bg)
            
        bg.logger.write(bg)

    bg.logger.close()

    elapsed_time = time() - start_time
    time_per_iteration = 1e3 * elapsed_time / bg.max_frame

    print('\nDone!')
    print(f'time/step: {time_per_iteration:10.2f} ms')
```

#### Choose density and temperature, then run simulation

```python
density     = 50    # kg / m^3
temperature = 145   # K
filename    = 'argon_rho%4.4d_T%4.4d.h5' % (density, temperature)

bag = define_initial_state(density, temperature, filename)

simulation_only = True

if simulation_only:
    simulate(bag)
else:
    build_scene(bag)
    sim = Sim(bag, update)
    sim.run()
    bag.logger.close()
```

```python
def radial_density_function(number_density, r, rmax, nbins):
    '''
    Compute the radial density function 

        g(r) = dC/dN.
        
    number_density : float   Average number of atoms per unit volume
    r : array [N, 3]         pPositions
    rmax : float             Maximum radial distance
    nbins :                  Number of bins so that Delta_r = rmax / nbins
    '''
    # Compute all pairwise displacements
    rij = r[:, None, :] - r[None, :, :]

    # Compute all pairwise distances, excluding zeros
    d = magnitude(rij)
    d = d[d > 0]   # Exclude zeros
    N = r.shape[0] # Number of particles

    # Histogram the distances
    edges = np.linspace(0.0, rmax, nbins + 1)
    counts, _ = np.histogram(d, bins=edges)

    # Convert counts to radial density function g(r)
    # -------------------------------------------------------
    # For an ideal gas in an unbounded volume the density is 
    # uniform, so the average number of radial distances in 
    # shells about every atom is given by
    #   N * n * shell_vol
    # where n =  number_density of the gas and shell_vol
    # are the volumes of the spherical shells
    
    rout, rin = edges[1:], edges[:-1] # Outer and inner radii of shells
    shell_vol = (4.0/3.0) * np.pi * (rout**3 - rin**3)
    ideal_gas_counts = N * number_density * shell_vol

    g  = counts / ideal_gas_counts
    rc = 0.5 * (rin + rout) # Radial distance to shell centers 
    
    return g, rc
```

```python
def maxwell_density_function(T, vmin, vmax, nbins):
    '''
    Compute the Maxwell velocity density function
    
        m(v) = df/dv.
    
    T : float          Dimensionless temperature: T = (1/3) <v^2>
    vmin, vmax : float Range of speeds
    nbins : int        Number of bins in speed
    '''
      
    v  = np.linspace(vmin, vmax, nbins+1)
    v  = 0.5*(v[1:]+v[:-1]) # centers of speed bins
    vv = v*v
    
    beta = 1/T  
    b  = 4*np.pi*np.sqrt((beta/(2*np.pi))**3)
    y  = b * vv * np.exp(-beta * vv/2)
    
    return y, v
```

```python
def read_sim_data(filename):
    
    with h5py.File(filename, 'r') as F:        
        r = np.array(F['r'])
        v = np.array(F['v'])
        U = np.array(F['U'])
        impulse = np.array(F['impulse'])
        
    return r, v, U, impulse
```
